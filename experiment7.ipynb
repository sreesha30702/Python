```python
#print all colors from colorset 1 not contained in colorset2 
set1={"red","black","yellow","blue"}
set2={"blue","white","orange"}
c=set1 - set2
print("colors in colorset1 but not in colorset2:",c)
```

```python
#print all colors from colorset 1 and colorset2 
set1={"red","black","yellow","blue"}
set2={"blue","white","orange"}
c=set1 | set2
print("all colors:",c)
```
