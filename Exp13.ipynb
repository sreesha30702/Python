```python
#module to calculate the area and perimeter of a Rectangle
%%writefile Rectangle.py
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
```

```python
#module to calculate the area and perimeter of a Circle
%%writefile Circle.py
import math
def area(r):
    return round(math.pi*r*r,2)
def perimeter(r):
    return round(2*math.pi*r,2)
```

```python
#Program to calculate area and perimeter of Rectangle and Circle using imported modules
import Rectangle
import Circle
print("Rectangle area:",Rectangle.area(10,5))
print("Rectangle perimeter:",Rectangle.perimeter(10,5))
print("Circle area:",Circle.area(7))
print("Circle perimeter:",Circle.perimeter(7))
```
