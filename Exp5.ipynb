```python
#write a python program to get 3rd element from last of a tuple
t=(10,20,30,40,50,60,70)
print("Enter 3rd element :",t[2])
print("Enter 3rd element from the last :",t[-3])
```

```python
#write a python program which accepts sequence of comma separated numbers from user and generate a tuple with those numbers and find average of all items in tuple
numbers = input("Enter numbers separated by commas :")
t = tuple(map(float,numbers.split(',')))
average = sum(t)/len(t)
print("Tuple:",t)
print("Average",average)
```
