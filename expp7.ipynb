```python
#write a p python program to printout all colors from color set 1 not contain in color set  2
s1=set(input("Enter colors for set 1)(separated by commas):").split(","))
s2=set(input("Enter colors for set 2)(separated by commas):").split(","))
print("set 1 :",s1)
print("set 2 :",s2)
print("Colos in set 1 not in set 2 : ",s1-s2)
```

```python
#write a python program to printout all colors from color set 1 andcolor set 2
s1=set(input("Enter colors for set 1)(separated by commas):").split(","))
s2=set(input("Enter colors for set 2)(separated by commas):").split(","))
print("set 1 :",s1)
print("set 2 :",s2)
print("Colos in set 1 and set 2 : ",s1|s2)
```
