```python
#GIVEN STRING TO NEW STRING WHERE FIRST AND LAST CHARS HAVE BEEN CHANGED
s=input("Enter a string:")
new_s=s[-1]+s[1:-1]+s[0]
print("New String=",new_s)
print(s[::-1])#reversing string
print(s[0:5:2])#prints from 0-4 by skipping one char
```

```python
#STRING MADE OF THE FIRST 2 AND LAST 2 CHARS FREOM A GIVEN STRING
s=input("Enter a string:")
fl_string=s[0:2]+s[-2:]
print("New String=",fl_string)
```

```python
#create a single string seperated with space from two strings by swapping the char at pposition
s1='hello'
s2 ='world'
a=s2[0]+s1[1:]+" "+s1[0]+s2[1:]
print("before swapping:",s1,s2)
print("swapped string:",a)
```

```python
#to get a string from a given string where all accurences of itsfirst char have been changed to $,expect first char itself
s=input("Enter a string:")
f=s[0]
rep=f+s[1:].replace(f,'$')
print("normal string:",s)
print("replaced by ''$':",rep)
```
